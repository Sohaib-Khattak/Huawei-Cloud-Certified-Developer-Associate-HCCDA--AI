TITANIC SURVIVAL ANALYSIS USING PANDAS

IMPORT THE NECESSARY LIBRARIES

```python
import pandas as pd
import seaborn as sns
import sklearn as sk
```

USED THE TITANIC DATASET

LOAD THE DATASET FROM THE SEABORN

```python
data=sns.load_dataset('titanic')
```

COMPLETE DATASET

```python
import pandas as pd

# Configure pandas to display all rows and columns without truncation
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(data)
```

CONVERT DATA TO CSV

```python
data.to_csv('titanic.csv')
```

```python
data
```

UNDERSTAND THE DATASET

```python
print(data.info())
```

#DATA CLEANING


```python
data.drop(columns=['deck'], inplace=True)
```

```python
data['age'].fillna(data['age'].median(), inplace=True)
```

```python
data['embarked'].fillna(data['embarked'].mode()[0], inplace=True)
```

```python
data['embark_town'].fillna(data['embarked'].mode()[0], inplace=True)
```

```python
data.isnull().sum()
```

```python
data
```

#DESCRIPTIVE ANALYSIS

```python
print("survival counts:\n", data['survived'].value_counts())
```

SURVIVAL RATE BY GENDER

```python
print("\nsurvival Rate By Gender:\n", data.groupby('sex')['survived'].mean())
```

SURVIVAL RATE BY CLASS

```python
print("\nsurvived Rate By Class:\n", data.groupby('class')['survived'].mean())
```

CROSS TAB OF GENEDR VS CLASS SURVIVAL

```python
pd.crosstab(data['sex'], data['pclass'], values=data['survived'], aggfunc='mean')
```

DATA VISUALIZATION WITH PANDAS

```python
import matplotlib.pyplot as plt

data.groupby('sex')['survived'].mean().plot(kind='bar', color=['blue', 'brown'], figsize=(6,3))

plt.title('survival Rate By Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate')
plt.ylim(0,1)
plt.grid(axis='y')
plt.show()
```

COUNTPLOT USING PANDAS

```python
data['pclass'].value_counts().sort_index().plot(kind='bar', color=['blue', 'brown', 'darkgreen'], figsize=(6,3))
plt.title('"Passanger count by class')
plt.xlabel('Class')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

SURVIVAL RATE BY EMBARKATION TOWN

```python
data.groupby('embark_town')['survived'].mean().plot(kind='bar', color=['maroon', 'gold', 'blue'], figsize=(6,3))
plt.title('Survival Rate By Embarkation Town')
plt.xlabel('Embarkation Town')
plt.ylabel('Survival Rate')
plt.ylim(0,1)
plt.grid(axis='y')
plt.show()
```
